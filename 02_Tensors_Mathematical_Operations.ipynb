3. Reduction Operation

```python
e = torch.randint(size=(2,3), low = 0, high = 10, dtype=torch.float32)
e
```

```python
# sum
torch.sum(e)
```

```python
# sum along columns
torch.sum(e, dim =0)
```

```python
# sum along rows
torch.sum(e, dim = 1)
```

```python
# mean
torch.mean(e)

# mean along columns
torch.mean(e, dim =0)

# mean along rows
torch.mean(e, dim = 1)
```

```python
# median
torch.median(e)
```

```python
# max and min
maxi = torch.max(e)
mini = torch.min(e)

print("MAX :", maxi)
print("MIN :", mini)
```

```python
# product
torch.prod(e)
```

```python
# standard deviation
torch.std(e)
```

```python
# variance
torch.var(e)
```

```python
# argmax -> position of max value
torch.argmax(e)
```

```python
# argmin -> position of min value
torch.argmin(e)
```

4. Matrix Operations

```python
f = torch.randint(size=(2,3), low = 0, high = 10)
g = torch.randint(size=(3,2), low = 0, high = 10)

print(f)
print(g)
```

```python
# matrix multiplication
torch.matmul(f, g)
```

```python
vector1 = torch.tensor([1, 2])
vector2 = torch.tensor([3, 4])

# dot product
torch.dot(vector1, vector2)
```

```python
# transpose
torch.transpose(f, 0 ,1)
```

```python
h = torch.randint(size=(3,3), low = 0, high = 10, dtype = torch.float32)
h
```

```python
# determinant
torch.det(h)
```

```python
# inverse
torch.inverse(h)
```

5. Comparison Operations

```python
i = torch.randint(size=(2,3), low = 0, high = 10)
j = torch.randint(size=(2,3), low = 0, high = 10)

print(i)
print(j)
```

```python
# greater than
i > j

# less than
i < j

# not equal to
i != j

# greater than equal to
i >= j

# lesser than equal to
i <= j
```

6. Special Functions

```python
k = torch.randint(size=(2,3), low = 0, high = 10, dtype = torch.float32)
k
```

```python
# log
torch.log(k)
```

```python
# exp
torch.exp(k)
```

```python
# sqrt
torch.sqrt(k)
```

```python
# sigmoid
torch.sigmoid(k)
```

```python
# softmax
torch.softmax(k, dim=0)
```

```python
# relu
torch.relu(k)
```

Inplace Operations

```python
m = torch.rand(2, 3)
n = torch.rand(2, 3)

print(m)
print(n)
```

```python
m.add_(n)
```

```python
torch.relu(m)
```

```python
# for inplace operations add "_" in the end -> it keeps changes permanent
m.relu_()
```
