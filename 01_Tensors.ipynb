```python
import torch
print(torch.__version__)
```

CREATING A TENSOR USING PYTORCH

```python
#   using empty
a = torch.empty(2,3)
```

```python
# check type
type(a)
```

```python
# using zeros
torch.zeros(2, 3)
```

```python
# using ones
torch.ones(2,3)
```

```python
# using rand -> gives random values
torch.rand(2,3)
```

```python
# use of seed -> to fetch same random values we initalize seed value
torch.manual_seed(100)
torch.rand(2,3)
```

```python
# using tensor
torch.tensor([[1,2,3],[4,5,6]])
```

```python
# other ways

# arange
print("using arange ->", torch.arange(0,12,2))

# using linspace -> linearly spaced
print("using linspace ->", torch.linspace(0,10,10))

# using eye -> identity matrix -> diagonal matrix
print("using eye ->", torch.eye(5))

# using full
print("using full ->", torch.full((3,3),5))
```

TENSOR SHAPES

```python
x = torch.tensor([[1,2,3],[4,5,6]])
x
```

```python
x.shape
```

```python
# to create a duplicate tensor from the existing shape
torch.empty_like(x)
```

```python
torch.zeros_like(x)
```

```python
torch.ones_like(x)
```

```python
torch.rand_like(x)
```

TENSORS DATA TYPES

```python
# find data types
x.dtype
```

```python
# assign data type
torch.tensor([1.0,2.0,3.0], dtype = torch.int32)
```

```python
torch.tensor([1,2,3], dtype = torch.float32)
```

```python
# using to()
x.to(torch.float32)
```
